```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
Year=[2017,2018,2019,2020,2021,2022]
Expense=[100000,120000,128000,110000,130000,150000]
Profit=[10000,20000,18000,21000,20000,27000]

startupx = {"Year":Year, "Expense":Expense, "Profit":Profit}
```

```python
df = pd.DataFrame(startupx)
df
```

# Line plot for Profit and Expense

```python
plt.figure(figsize=(8, 6))
plt.plot(df["Year"], df["Expense"], label='Expense', color='blue', marker='o')
plt.plot(df["Year"], df["Profit"], label='Profit', color='green', marker='s')
plt.title("Profit and Expense Over Years")
plt.xlabel("Year")
plt.ylabel("Amount (EGP)")
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```

# Bar Chart - Expense per Year

```python
plt.figure(figsize=(8, 6))
plt.bar(df["Year"], df["Expense"], color='skyblue')
plt.title("Annual Expense Overview")
plt.xlabel("Year")
plt.ylabel("Expense (EGP)")
plt.show()
```

# Scatter Plot for Profit

```python
plt.figure(figsize=(8, 6))
plt.scatter(df["Year"], df["Profit"], color="purple", label="Profit", s=100)
plt.title("Profit Over Years")
plt.xlabel("Year")
plt.ylabel("Profit (EGP)")
plt.grid(True)
plt.legend()
plt.show()
```

# Scatter Plot for Expense vs Profit

```python
plt.figure(figsize=(8, 6))
plt.scatter(df["Expense"], df["Profit"], color="red", label="Expense vs Profit", s=100)
plt.title("Expense vs Profit Correlation")
plt.xlabel("Expense (EGP)")
plt.ylabel("Profit (EGP)")
plt.grid(True)
plt.legend()
plt.show()
```
